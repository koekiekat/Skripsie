Imports

```python

import matplotlib.pyplot as plt
import numpy as np

from Background_functions import read_audio_file, split_audio_segments, start_end_times
from STFT import short_time_calc, plot_spectrogram
from DFT import dtw_calc

from pathlib import Path

DATA_22 = Path(r"C:\Users\HP\Desktop\Skripsie data\DataSubmission\2023_04_22")
```

Data download

```python
audio = DATA_22 / "20230422_171301.WAV"
f_s, x = read_audio_file(audio)
```

Template

```python
text = DATA_22 / "20230422_171301.Detections.selections.txt"
temp_marks, start_t_temp, end_t_temp = start_end_times(text)

whale_call_segments = split_audio_segments(x, f_s, start_t_temp, end_t_temp)
num_temps =len(whale_call_segments)
print(f"Number of calls in given file: {num_temps}")
audio_template_1 = whale_call_segments[14]
```

Running loop for STFT template analysis

```python
tot_calls = 259 
stft_cost = np.zeros(num_temps)
f_temp, t_temp, Zxx_temp, fs_new = short_time_calc(audio_template_1, f_s)

for i in range(num_temps):
    _, _, Zxx_seg, _=short_time_calc(whale_call_segments[i], f_s)
    stft_cost[i] = dtw_calc(Zxx_temp, Zxx_seg)

sorted_idx = np.argsort(stft_cost)
single_tone_idx = sorted_idx[:tot_calls]
single_tone_costs = stft_cost[single_tone_idx]

print(f"Cost range among {tot_calls} candidate single tones: " f"{single_tone_costs.min():.4f} - {single_tone_costs.max():.4f}")
```

```python
# --- Interactive selection loop: single y/n, auto-split into templates/calibration ---
n_templates = 25
n_calibration = 155
n_total = n_templates + n_calibration

print("Spectrogram of Template:")
plot_spectrogram(f_temp, t_temp, Zxx_temp, fs_new)

approved_idx = []      # order of approval, in rank order
rejected_idx = set()

candidate_ranks = np.unique(np.linspace(0, len(single_tone_idx) - 1, len(single_tone_idx)).round().astype(int))

for rank in candidate_ranks:
    if len(approved_idx) >= n_total:
        break

    idx = int(single_tone_idx[rank])
    if idx in rejected_idx or idx in approved_idx:
        continue

    f_sig, t_sig, Zxx_sig, fs_new = short_time_calc(whale_call_segments[idx], f_s)
    plot_spectrogram(f_sig, t_sig, Zxx_sig, fs_new)
    plt.show()
    print(f"rank: {rank}/{len(single_tone_idx)-1} | dtw cost: {stft_cost[idx]:.4f} | "
          f"index: {idx} | start_t: {start_t_temp[idx]}")
    print(f"approved so far: {len(approved_idx)}/{n_total}")

    resp = input("Keep this call? (y/n/stop): ").strip().lower()
    if resp == "y":
        approved_idx.append(idx)
    elif resp == "stop":
        break
    else:
        rejected_idx.add(idx)

# --- Split approved calls proportionally into templates vs calibration ---
# Both buckets get every ~n_total/n_templates-th approved call, so each
# bucket spans the full best->worst range rather than one grabbing early picks.
template_slots = np.linspace(0, len(approved_idx) - 1, min(n_templates, len(approved_idx)))
template_slots = np.unique(template_slots.round().astype(int))

template_idx = [approved_idx[i] for i in template_slots]
calibration_idx = [approved_idx[i] for i in range(len(approved_idx)) if i not in template_slots]

print(f"\nTemplates ({len(template_idx)}): {template_idx}")
print(f"Calibration ({len(calibration_idx)}): {calibration_idx}")
```
